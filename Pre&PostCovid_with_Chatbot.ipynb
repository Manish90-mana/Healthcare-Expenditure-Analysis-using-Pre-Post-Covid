### .dat → raw fixed-width ASCII data
###.sas → layout file describing column positions

```python
import pandas as pd
import re
import numpy as np

def read_meps_ascii(dat_file, sas_file):## It converts the ascii to dataframe
    colspecs = []
    names = []

    with open(sas_file, 'r') as file:
        for line in file:
            match = re.search(r'@(\d+)\s+(\w+)\s+\$?(\d+)\.', line)
            if match:
                start = int(match.group(1)) - 1
                width = int(match.group(3))
                colspecs.append((start, start + width))
                names.append(match.group(2))

    df = pd.read_fwf(dat_file, colspecs=colspecs, names=names)

    # Replace MEPS special missing codes
    df.replace([-1, -7, -8, -9], np.nan, inplace=True)

    return df
```

1.It reads the .sas file

2.it extracts column start position,column width,variable name

3.It replaces special MEPS missing code ex:[-1,-7,-8,-9] to NAN  

4.this function converts meps raw data files into Pandas dataframe

```python
import joblib

model = joblib.load("rf_model.pkl")
medians = joblib.load("medians.pkl")
columns = joblib.load("columns.pkl")

print(" Model loaded successfully")
```

```python
def predict_user(input_data):
    df = pd.DataFrame([input_data])
    for col in medians:
        if col not in df:
            df[col] = medians[col]

    df = df.fillna(medians)
    df = pd.get_dummies(df)
    df = df.reindex(columns=columns, fill_value=0)
    prob = model.predict_proba(df)[0][1]
    pred = 1 if prob >= 0.5 else 0

    return pred, prob
```

```python
def explain(input_data):
    reasons = []
    if input_data["POVCAT"] >= 4:
        reasons.append("Belongs to high poverty category (financial stress)")

    elif input_data["POVCAT"] == 1:
        reasons.append("Low poverty level (financially stable)")
    if input_data["RXEXP"] > 10000:
        reasons.append("Very high medical expenses")

    elif input_data["RXEXP"] > 3000:
        reasons.append("Moderate medical expenses")

    if input_data["AGE"] > 60:
        reasons.append("Older age increases healthcare needs")

    elif input_data["AGE"] < 18:
        reasons.append("Young age (generally lower healthcare burden)")
    if input_data["INSURC"] == 3:
        reasons.append("No insurance (high financial burden risk)")

    elif input_data["INSURC"] == 1:
        reasons.append("Has private insurance (reduces burden risk)")
    if input_data["EMPST_FINAL"] == 0:
        reasons.append("Unemployed (financial instability)")

    return reasons
```

```python
def chatbot():
    print("\n Healthcare Burden Prediction Chatbot")
    print("Enter patient details:\n")

    age = int(input("Age: "))
    rxexp = float(input("Prescription Expense: "))

    insurc = int(input(
        "Insurance Category:\n"
        "1 = Private Insurance\n"
        "2 = Public Insurance (Medicare/Medicaid)\n"
        "3 = Uninsured\n"
        "Enter choice: "
    ))

    povcat = int(input("Poverty Category (1–5): "))
    sex = int(input("Sex (1=Male, 2=Female): "))
    race = int(input("Race (1–6): "))
    emp = int(input("Employment (0=Unemployed, 1=Employed): "))

    input_data = {
        "AGE": age,
        "RXEXP": rxexp,
        "INSURC": insurc,
        "POVCAT": povcat,
        "SEX": sex,
        "RACEV1X": race,
        "EMPST_FINAL": emp,
        "PROBPY_final": 0,
        "CRFMPY_final": 0,
        "PYUNBL_final": 0
    }

    pred, prob = predict_user(input_data)

    if prob >= 0.7:
        risk_level = " HIGH RISk"
    elif prob >= 0.4:
        risk_level = " MEDIUM RISK"
    else:
        risk_level = "LOW RISK"

    print("\n RESULT:")
    print(f"{risk_level} (Probability: {prob:.2f})")

    reasons = explain(input_data)
    print("\n Explanation:")
    if reasons:
        for r in reasons:
            print("-", r)
    else:
        print("No major risk factors detected.")

    tips = []

    if rxexp > 5000:
        tips.append("Consider insurance plans to reduce medical expenses")

    if insurc == 3:
        tips.append("Apply for public/government insurance schemes")

    if povcat >= 4:
        tips.append("Check eligibility for financial assistance programs")

    if age > 60:
        tips.append("Regular health checkups recommended")

    if emp == 0:
        tips.append("Explore employment or financial support options")

    print("\n Recommendations:")
    if tips:
        for t in tips:
            print("-", t)
    else:
        print("No immediate recommendations needed.")
```
